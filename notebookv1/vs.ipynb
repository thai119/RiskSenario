```python
import pandas as pd

# Đọc file CSV
file_path = "D:/Downloads/csv/k2.pcap_Flow.csv"  
# Đổi thành đường dẫn file thực tế
df = pd.read_csv(file_path)

# Tạo mapping giữa tên cột cũ (features_1) và tên cột mới (features_2)
column_mapping = {
    "Flow Duration": "bidirectional_duration_ms",
    "Tot Fwd Pkts": "src2dst_packets",
    "Tot Bwd Pkts": "dst2src_packets",
    "TotLen Fwd Pkts": "src2dst_bytes",
    "TotLen Bwd Pkts": "dst2src_bytes",
    "Flow IAT Mean": "bidirectional_mean_piat_ms",
    "Flow IAT Std": "bidirectional_stddev_piat_ms",
    "Flow IAT Max": "bidirectional_max_piat_ms",
    "Flow IAT Min": "bidirectional_min_piat_ms",
    "Fwd IAT Mean": "src2dst_mean_piat_ms",
    "Fwd IAT Std": "src2dst_stddev_piat_ms",
    "Fwd IAT Max": "src2dst_max_piat_ms",
    "Fwd IAT Min": "src2dst_min_piat_ms",
    "Bwd IAT Mean": "dst2src_mean_piat_ms",
    "Bwd IAT Std": "dst2src_stddev_piat_ms",
    "Bwd IAT Max": "dst2src_max_piat_ms",
    "Bwd IAT Min": "dst2src_min_piat_ms",
    "FIN Flag Cnt": "bidirectional_fin_packets",
    "SYN Flag Cnt": "bidirectional_syn_packets",
    "RST Flag Cnt": "bidirectional_rst_packets",
    "PSH Flag Cnt": "bidirectional_psh_packets",
    "ACK Flag Cnt": "bidirectional_ack_packets",
    "URG Flag Cnt": "bidirectional_urg_packets",
    "CWE Flag Count": "bidirectional_cwr_packets",
    "ECE Flag Cnt": "bidirectional_ece_packets"
}

# Đổi tên cột theo mapping
df.rename(columns=column_mapping, inplace=True)

# Chỉ giữ lại các cột đã được đổi tên
selected_columns = list(column_mapping.values())  # Lấy danh sách cột mới
df = df[selected_columns]

# Hiển thị vài dòng đầu của DataFrame sau khi lọc
print(df.head())

# Lưu lại file CSV với cột đã đổi tên (nếu cần)
df.to_csv("D:/Downloads/data/k2.pcap_Flow.csv", index=False)
```
