```python
import pandas as pd
df_aa = pd.read_csv('D:/Downloads/data apt/AA_combined.csv', low_memory=False)
df_apt = pd.read_csv('D:/Downloads/data apt/APT_combined.csv', low_memory=False)
```

```python
# Gộp hai DataFrame theo chiều dọc (nối các hàng)
df_combined = pd.concat([df_aa, df_apt], ignore_index=True)

# Lưu file gộp (tùy chọn)
df_combined.to_csv('D:/Downloads/data apt/merge.csv', index=False)

# Xem trước dữ liệu gộp (tùy chọn)
print(df_combined.head())
```
